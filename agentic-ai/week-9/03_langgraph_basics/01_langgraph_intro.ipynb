# Lesson 01 — Introduction to LangGraph

## Objective
Understand how LangGraph builds workflows using nodes, edges, and shared state.

## Workflow
START → Analyze Node → Response Node → END

## Core Concepts
- State: Shared information carried throughout the workflow.
- Node: An individual task performed by a function.
- Edge: Defines the execution sequence.
- Graph: The complete connected workflow.

```python
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END

print("LangGraph imports successful!")
```


## 1. Define the State

State is the shared data structure passed between nodes.

Our workflow will store:
- query: Original user question
- result: Current/final result

```python
class AgentState(TypedDict):
    query: str
    result: str
```

## 2. Create Nodes

Each node is a Python function that:
1. Receives the current state.
2. Performs a task.
3. Returns a dictionary containing state updates.

```python
def analyze_node(state: AgentState):
    print("Analyzing query:", state["query"])

    return {
        "result": "25% of 4800"
    }


def response_node(state: AgentState):
    print("Generating response...")

    return {
        "result": "25% of 4800 = 1200"
    }
```

## 3. Build the Graph

StateGraph creates our workflow.

Nodes represent tasks.
Edges define execution order.
START and END define workflow boundaries.

```python
builder = StateGraph(AgentState)

# Register nodes
builder.add_node("analyze", analyze_node)
builder.add_node("response", response_node)

# Connect nodes
builder.add_edge(START, "analyze")
builder.add_edge("analyze", "response")
builder.add_edge("response", END)

# Compile graph
graph = builder.compile()

print("Graph compiled successfully!")
```

```python
initial_state = {
    "query": "Calculate 25% of 4800",
    "result": ""
}

final_state = graph.invoke(initial_state)

print("\nFinal Output:")
print(final_state)
```
