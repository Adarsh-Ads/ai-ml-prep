# Lesson 03 — Nodes in Depth

## Objective
Understand node responsibilities, inputs, outputs, and execution order.

## Workflow
START → Clean Text → Analyze Text → Generate Response → END

```python
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END

class TextState(TypedDict):
    text: str
    cleaned_text: str
    word_count: int
    response: str
```

```python
def clean_text(state: TextState):
    cleaned = state["text"].strip().lower()

    print("Cleaning text...")

    return {"cleaned_text": cleaned}
```

```python
def analyze_text(state: TextState):
    count = len(state["cleaned_text"].split())

    print("Analyzing text...")

    return {"word_count": count}
```

```python
def generate_response(state: TextState):
    response = (
        f"Your text contains {state['word_count']} words."
    )

    print("Generating response...")

    return {"response": response}
```

```python
builder = StateGraph(TextState)

builder.add_node("clean", clean_text)
builder.add_node("analyze", analyze_text)
builder.add_node("respond", generate_response)

builder.add_edge(START, "clean")
builder.add_edge("clean", "analyze")
builder.add_edge("analyze", "respond")
builder.add_edge("respond", END)

graph = builder.compile()
```

```python
initial_state = {
    "text": "   LangGraph makes AI workflows easier to manage.   ",
    "cleaned_text": "",
    "word_count": 0,
    "response": ""
}

final_state = graph.invoke(initial_state)

print("\nFinal Response:")
print(final_state["response"])
```

```python
for step in graph.stream(initial_state, stream_mode="updates"):
    print(step)
```
