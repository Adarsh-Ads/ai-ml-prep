# Lesson 04 — Edges in LangGraph

## Objective
Understand how edges control the execution sequence of nodes.

## Workflow
START → Read → Process → Generate → END

```python
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END

class WorkflowState(TypedDict):
    message: str
    processed_message: str
    final_output: str
```

```python
def read_node(state: WorkflowState):
    print("Node A: Reading input")

    return {
        "processed_message": state["message"].strip()
    }
```

```python
def process_node(state: WorkflowState):
    print("Node B: Processing input")

    processed = state["processed_message"].upper()

    return {
        "processed_message": processed
    }
```

```python
def generate_node(state: WorkflowState):
    print("Node C: Generating output")

    return {
        "final_output": f"Processed message: {state['processed_message']}"
    }
```

```python
builder = StateGraph(WorkflowState)

builder.add_node("read", read_node)
builder.add_node("process", process_node)
builder.add_node("generate", generate_node)

# Define execution sequence
builder.add_edge(START, "read")
builder.add_edge("read", "process")
builder.add_edge("process", "generate")
builder.add_edge("generate", END)

graph = builder.compile()

print("Graph compiled successfully!")
```

```python
initial_state = {
    "message": "   learning langgraph is useful   ",
    "processed_message": "",
    "final_output": ""
}

final_state = graph.invoke(initial_state)

print("\nFinal Output:")
print(final_state["final_output"])
```
